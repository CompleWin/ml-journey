```python
import matplotlib.pyplot as plt
import numpy as np
from sympy.printing.pretty.pretty_symbology import line_width
```

```python
x = [1, 2, 3, 4, 5, 6]
y = [1, 4, 9, 16, 25, 36]

plt.plot(x, y)
plt.grid() # Сетка на графике
plt.xlabel('x')
plt.ylabel('y')
plt.title('$y=x^2$')
plt.show()
```

Диаграмма рассеяния. Отдельные точки при помощи функции `scatter`

```python
x = [1, 2, 3, 4, 5, 6]
y = [1, 4, 9, 16, 25, 36]

plt.scatter(x,y)
plt.grid() # Сетка на графике
plt.xlabel('x')
plt.ylabel('y')
plt.title('$y=x^2$')
plt.show()
```

Несколько диаграмм на одном графике

```python
x = [1, 2, 3, 4, 5]
y1 = [1, 4, 9, 16, 25]
y2 = [2, 4, 6, 8, 10]

plt.scatter(x, y1, label='$y2$')
plt.scatter(x, y2, label='$y1$')
plt.legend()
plt.show()
```

Графики могут быть по разным количествам точек

```python
x1 = [1, 2, 3, 4, 5]
y1 = [1, 4, 9, 16, 25]
x2 = [1, 5]
y2 = [2, 10]

plt.plot(x1, y1, label='$y1$')
plt.plot(x2, y2, label='$y2$')
plt.legend()
plt.show()
```

Цвета графикам назначаются автоматически, но его можно указать вручную, с помощью параметра `color=`

```python
x1 = [1, 2, 3, 4, 5]
y1 = [1, 4, 9, 16, 25]
x2 = [1, 5]
y2 = [2, 10]

plt.plot(x1, y1, label='$y1$', color='red')
plt.plot(x2, y2, label='$y2$', color='blue')
plt.legend()
plt.show()
```

### Горизонтальные и вертикальные линии на графике
`axvline` -- вертикальная линия
`axhline` -- горизонтальная линия

```python
x = np.linspace(0.2, 2)
y1 = x ** 2
y2 = 1 / x
plt.plot(x, y1, label='$y=x^2$')
plt.plot(x, y2, label='$y=\frac{1}{x}$')
plt.axhline(1, color='black')
plt.axvline(1, color='black')
plt.show()
```

### Несколько графиков

`plt.subplot(rows, columns, index)` \
rows -- количество строк \
columns -- количество колонок \
index -- порядковый номер графика, с которым мы сейчас будем работать. Начинается с 1, а не с 0

```python
plt.subplot(1, 2, 1)
plt.plot([0, 1], [0, 1])
plt.subplot(1, 2, 2)
plt.plot([0, 1], [2, 4])
plt.show()
```

Друг под другом

```python
plt.subplot(2, 1, 1)
plt.plot([0, 1], [0, 1])
plt.subplot(2, 1, 2)
plt.plot([0, 1], [2, 4])
plt.show()
```

### Кастомизация графиков

`linewidth` -- ширина линии (по умолчанию = 1) \
`linestyle` -- стиль линии (по умолчанию = solid), варианты: dashed, dotted, dashdot \

#### Маркеры
`marker` -- каждую точку на линии можно отметить определенным маркером: o - круг, v - треугольник, d - ромб \
`marker-size` -- размер маркера (у `scatter` размер маркера это параметр `s`) \
`c` -- цвет маркера у `scatter`, в отличие от `color` может принимать список цветов

`alpha` -- управляет прозрачностью графика


```python
x = [1, 2, 3]
y1 = [1, 2, 3]
y2 = [2, 3, 4]
y3 = [3, 4, 5]

plt.plot(x, y1, linewidth=2, linestyle='dashed', label='linewidth=2; linestyle=dashed')
plt.plot(x, y2, linewidth=3, linestyle='dotted', label='linewidth=3; linestyle=dotted')
plt.plot(x, y3, linewidth=4, linestyle='dashdot', label='linewidth=4; linestyle=dashdot')
plt.legend()
plt.show()
```

```python
x = [1, 2, 3]
y1 = [1, 2, 3]
y2 = [2, 3, 4]
y3 = [3, 4, 5]

plt.plot(x, y1, marker='o', label='o - круг')
plt.plot(x, y2, marker='v', label='v - треугольник')
plt.plot(x, y3, marker='d', label='d - ромб')
plt.legend()
plt.show()
```

## Пределы на графике

`xlim` -- предел по x \
`ylim` -- предел по y \
`xticks` -- количество отрезков по оси x \
`yticks` -- количество отрезков по оси y \
В `xticks` и `yticks` можно передавать пустой список, тогда отметки на графике пропадут. Удобно для вывода изображений

```python
x = [1, 2, 3]
y = [1, 2, 3]
plt.plot(x, y)
plt.xlim(1, 3)
plt.ylim(1, 3)
plt.xticks([1, 2, 3])
plt.yticks([1, 2, 3])
plt.show()
```

```python
x = [1, 2, 3]
y = [1, 2, 3]
plt.plot(x, y)
plt.xlim(1, 3)
plt.ylim(1, 3)
plt.xticks([1, 2, 3], ["one", "two", "three"])
plt.yticks([1, 2, 3], ["one", "two", "three"])
plt.show()
```
