# NumPy
Это библиотека для эффективной работе с массивами, в том числе многомерными, начиная с базовых математических операций и заканчивая линейной алгеброй

```python
import numpy as np
```

## Способы создания списков

```python
a = np.array([1, 2, 3]) # Используя обычный список
b = np.array([[1, 2, 3], [4, 5, 6]])

c = np.zeros(5) # массив нулей
d = np.ones(7) # массив единиц

# можно передавать размерность
e = np.zeros((3, 2))
print(e)
```

Массив последовательных чисел можно создать с помощью функции `arange` это сокращение от array и range. Аргументы те же, что и у встроенной функции `range`
> В отличие от `range`, создаться массив, а не итератор. Это может занять много памяти.

```python
a = np.arange(5)
b = np.arange(1, 10, 2)
print(a)
print(b)
```

```python
# начальная и конечная точка (включительно) отрезка и количество между ними
a = np.linspace(0, 5, 50)
print(a)
```

## Рандом в NumPy (пакет random)

```python
a = np.random.randint(1, 10, 5) # 5 случайных чисел от 1 (вкл) до 10 (невкл)
print(a)

# вместо количество можно передать размерность
a = np.random.randint(1, 10, (2, 3))
print(a)
```

Преимущество массив в том, что они созданы специально под математические операции над ними \
Например в NumPy с массивами можно делать все то, что можно с числами. Тогда эти операции будут применяться поэлементно

```python
a = np.array([1, 2, 3])
b = np.array([3, 5, 6])
c = a - b
print(c)
```

Массивы в NumPy в разы быстрее обычных списков в Python

```python
import time
a = np.arange(10_000_000)
b = np.arange(10_000_000)
start = time.time()
c = a - b
print(f'{time.time() - start:.4f}s')
```

В NumPy можно делить на ноль, результат будет бесконечность, но мы получим RuntimeWarning

```python
a = np.array([1])
b = np.array([0])
print(a / b)
```

Но в NumPy нельзя делить ноль на ноль и брать корень из отрицательного числа, ошибки не будет, но результат будет nan -- Not a number. \
При это nan это не None из Python, чтобы проверить на nan нужно использовать `np.isnan`

```python
a = np.array([-1]) ** 0.5
b = np.array([0]) / np.array([0])
print(a)
print(b)

print(a is None) # False
print(np.isnan(a))
```

## Линейная алгебра в NumPy

Скалярное произведение векторов это функция `dot` \
Матричное умножение это оператор @

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print(np.dot(a, b))
```

Задача #1 - проверить основное тригонометрическое тождество: сумма квадрата синуса и квадрата косинуса любого угла должна быть равна единице.

```python
a = np.arange(5)

print(np.sin(a) ** 2 + np.cos(a) ** 2)
```

`shape` показывает размерность массива, она всегда кортеж, даже для списков \
`size` показывает количество элементов в массиве

```python
a = np.array([1, 2, 3])
b = np.array([
        [1, 2, 3],
        [4, 5, 6]
])

print(a.shape, a.size)
print(b.shape, b.size)
```

`reshape` меняет размерность массива. Эта функция, как и другие в numpy не изменяет массив, а создает новый

```python
a = np.arange(1, 7)
print(a)
print(a.shape)

print(a.reshape((3, 2)))
```

Чтобы узнать тип данных массива в NumPy можно использовать метод `.dtype`

```python
a = np.array([1, 2, 3])
b = np.array([1.2, 3.4, 1.2])

print(a.dtype)
print(b.dtype)
```

Привести массив к определенному типу данных можно с помощью `astype` \
При переводе из вещественной в целочисленный все числа округляются вниз

```python
a = a.astype(np.float64)
print(a)
print(a.dtype)
```

```python
a = np.array([1.2, 3.49, 3.8])
a = a.round() # округлит до ближайшего целого, тип при этом не изменится
print(a)
```

```python
a = np.array([6, 7, 1, 2])
print(a.min())
print(a.max())
print(a.sum())
```

Для вычисления среднего в NumPy есть `.mean` \
Для вычисления стандартного отклонения `.std`

```python
a = np.array([6, 7, 1, 2])
print(a.mean())
print(a.std())
```

Задача #2 - нормализовать массив случайных чисел: привести все элементы к диапазону [0; 1], а потом сделать обратное преобразование.

Min-Max нормализация (линейное масштабирование)

```python
a = np.random.randint(1, 100, 10)
print(a)

x_min = a.min()
x_maxmin = a.max() - a.min()

a_new = (a - x_min) / x_maxmin
print(a_new)

a_old = (a_new * x_maxmin + x_min).astype(np.int64)
print(a_old)
```

Маска по логическим функциям \
Вместо or, and, not в numpy надо писать |, &, ~

```python
a = np.array([1, 2, 3, 4, 5, 6, 7, 8])
print(a[a % 2 == 0])
print(a[(a % 2 == 0) | (a % 3 == 0)])
```

Задача #3 - убрать выбросы из массива: удалить все элементы, которые больше среднего в 3 раза или меньше 1/3 от среднего.

```python
a = np.random.randint(1, 100, 20)
print(a)

mean = a.mean()
filtered = a[(a < mean * 3) & (a > mean / 3)]
print(filtered)
```
